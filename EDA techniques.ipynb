# Missing value treatment

```python
clean_data=emp.copy()
```

```python
import numpy as np
```

# working on numeric values which are missing

```python
clean_data["Age"]=clean_data["Age"].fillna(np.mean(pd.to_numeric(clean_data["Age"])))   # fillull-->to fill missing values
```

```python
clean_data["Age"]
```

```python
clean_data["Exp"]=clean_data["Exp"].fillna(np.mean(pd.to_numeric(clean_data["Exp"])))
```

```python
clean_data["Exp"]
```

```python
clean_data["Location"]=clean_data["Location"].fillna(clean_data["Location"].mode()[0])  # when ever we categorical we have give zero right side
```

```python
clean_data["Location"]
```

```python
clean_data
```

```python
clean_data["Age"]=clean_data['Age'].astype(int)   # covert float to int in age
```

```python
clean_data["Age"]
```

```python
clean_data["Exp"]=clean_data['Exp'].astype(int) 
clean_data["Exp"]
```

```python
clean_data["Salary"]=clean_data['Salary'].astype(int) 
clean_data["Salary"]
```

```python
clean_data.info()
```

```python
clean_data["Name"]=clean_data['Name'].astype('category')
clean_data['Name']
```

```python
clean_data["Location"]=clean_data['Location'].astype('category') 
clean_data["Location"]
```

```python
clean_data["Domain"]=clean_data['Domain'].astype('category') 
clean_data['Domain']
```
